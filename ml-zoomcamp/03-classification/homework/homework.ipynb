```python
# load data as dataframe df
df = pd.read_csv('https://raw.githubusercontent.com/alexeygrigorev/datasets/master/course_lead_scoring.csv')

"""
df.shape()
df.info()
df.head()
df.nunique()
df.dtypes
"""


# Data Preparation, check missing values
df.isna().sum()

# Managing missing values
# Separate categorical and numerical columns
categorical = df.select_dtypes(include=['object', 'string']).columns.tolist()
numerical = df.select_dtypes(include='number').columns.tolist()

# Fill missing values - categorical 'NA', numerical 0.0
df[categorical] = df[categorical].fillna('NA')
df[numerical] = df[numerical].fillna(0.0)

#  remove target `converted` from numerical
numerical = [col for col in numerical if col != 'converted']
```
